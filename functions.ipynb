# **oop**

```python
class Car:  # Class definition
    def __init__(self, brand, model, year):  # Constructor
        self.brand = brand  # Attribute
        self.model = model
        self.year = year

    def display_info(self):  # Method
        print(f"Car: {self.brand} {self.model}, Year: {self.year}")

# Creating an object (instance)
my_car = Car("Toyota", "Corolla", 2022)

# Calling a method
my_car.display_info()
```

### **Inheritance**

# **1. Single Inheritance**

```python
# Parent class
class Animal:
    def speak(self):
        print("Animal makes a sound")

# Child class (inherits from Animal)
class Dog(Animal):
    def bark(self):
        print("Dog barks")

# Creating an object of Dog
dog = Dog()
dog.speak()  # Inherited from Animal
dog.bark()   # Defined in Dog
```

# **2. Multiple Inheritance**

```python

class Father:
    def work(self):
        print("Father works as an engineer")

class Mother:
    def cook(self):
        print("Mother cooks food")


class Child(Father, Mother):
    def play(self):
        print("Child loves to play")


c = Child()
c.work()
c.cook()
c.play()
```

# ***Multilevel Inheritance***

```python

class Grandparent:
    def family_name(self):
        print("Family name: eruvaram")


class Parent(Grandparent):
    def surname(self):
        print("Surname: gireesh")

class Child(Parent):
    def name(self):
        print("First Name: eruvaram")

# Creating an object of Child
c = Child()
c.family_name()
c.surname()
c.name()
```

# **Polymorphism (Multiple Forms)**
Polymorphism allows different classes to define the same method name, but each has a different behavior.

```python
class Animal:
    def make_sound(self):
        print("Animal makes a sound")

class Dog(Animal):
    def make_sound(self):
        print("boow")

class Cat(Animal):
    def make_sound(self):
        print("Meow")

# Function using polymorphism
def sound_test(animal):
    animal.make_sound()

dog = Dog()
cat = Cat()

sound_test(dog)
sound_test(cat)
```

# **Encapsulation
Encapsulation is the concept of restricting direct access to some data inside a class, allowing controlled access through methods.

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private variable

    # Getter method
    def get_balance(self):
        return self.__balance

    # Setter method
    def set_balance(self, amount):
        if amount < 0:
            print("Invalid balance")
        else:
            self.__balance = amount

# Creating object
acc = BankAccount(1000)

# Accessing balance via getter
print(acc.get_balance())  # ✅ 1000

# Modifying balance using setter
acc.set_balance(2000)
print(acc.get_balance())  # ✅ 2000

# Trying to set a negative balance
acc.set_balance(-500)  # ❌ Invalid balance
```

# **Abstraction**
Abstraction is the concept of hiding unnecessary details and only exposing essential features.

In Python, abstraction is implemented using Abstract Base Classes (ABC).

```python
from abc import ABC, abstractmethod

# Abstract class
class Vehicle(ABC):
    @abstractmethod
    def start(self):
        pass  # No implementation in abstract class

    @abstractmethod
    def stop(self):
        pass

# Concrete class (Child class implementing abstract methods)
class Car(Vehicle):
    def start(self):
        print("Car engine started!")

    def stop(self):
        print("Car stopped!")

# Creating an object of concrete class
my_car = Car()
my_car.start()
my_car.stop()
```
