# **Variables**
Rules for Naming Variables



To use variables effectively, we must follow Python’s naming rules:

Variable names can only contain letters, digits and underscores (_).
A variable name cannot start with a digit.
Variable names are case-sensitive (myVar and myvar are different).
Avoid using Python keywords (e.g., if, else, for) as variable names.



```python
# Variable 'x' stores the integer value 10
x = 5

# Variable 'name' stores the string "Samantha"
name = "Samantha"

print(x)
print(name)
```

```python
# Basic Assignment
x = 5
y = 3.14
z = "Hi"
print (x,y,z)
```

```python
# Dynamic Typing
x = 10
x = "Now a string"
print (x)
#Assigning the Same Value

a = b = c = 100
print(a, b, c)
# Assigning Different Values
g, i, r = 1, 2.5, "Python"
print(g, i, r)
```

# **Data structures**

# **1. Lists (list)**
Ordered, mutable (modifiable), and allows duplicate values.
Elements can be accessed by index.
Can store different data types in the same list.

```python
my_list = [1, 2, 3, "hello", 3.5]
my_list.append(4)  # Adds an element
my_list.remove(2)  # Removes the element 2
print(my_list[0])  # Access first element
```

```python
thislist = ["apple", "banana", "cherry"]
print(thislist[-1])
thislist[1] = "blackcurrant"
print(thislist)
```

```python
fruits = ['apple', 'banana', 'cherry', 'orange']

fruits.clear()
print(fruits)
```

```python
fruits = ["apple", "banana", "cherry"]

x = fruits.copy()

print(x)
```

```python
#extend() Method
fruits = ['apple', 'banana', 'cherry']

cars = ['Ford', 'BMW', 'Volvo']

fruits.extend(cars)
print(fruits)
# reverse() Method
cars.reverse()
print (cars)
```

# **2. Tuples (tuple)**



1.   Ordered, immutable (cannot be modified), and allows duplicates.

2. Faster than lists due to immutability.





```python
my_tuple = (1, 2, 3, "hello", 3.5)
print(my_tuple[1])  # Accessing elements
```

```python
thistuple = (1, 3, 7, 8, 7, 5, 4, 6, 8, 5)

x = thistuple.count(5)

print(x)
```

# **3. Sets (set)**
Unordered, mutable, and does not allow


duplicate values.
Useful for removing duplicates and performing set operations like union, intersection, etc.


```python
my_set = {1, 2, 3, 4, 4, 5}
my_set.add(6)  # Adds an element
my_set.remove(2)  # Removes an element
print(my_set)  # Output: {1, 3, 4, 5, 6}
```

```python
# difference() Method
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "apple"}

z = x.difference(y)

print(z)


a = {"apple", "banana", "cherry"}
b = {"google", "microsoft", "apple"}
c = {"cherry", "micra", "bluebird"}
myset = a - b

print(myset)
myset2 = a.difference(b, c)

print(myset2)
```

# **4. Dictionaries (dict)**

1.   Unordered (Python 3.6 and earlier), but ordered from Python 3.7+.

2.  Stores data in key-value pairs.
3.Keys must be unique and immutable (e.g., strings, numbers, or tuples).




```python
thisdict = {
  "brand": "Ford",
  "model": "Mustang",
  "year": 1964
}
print(thisdict["brand"])
```

```python
my_dict = {"name": "Alice", "age": 25, "city": "New York"}
print(my_dict["name"])  # Access value using key
my_dict["age"] = 26  # Modifying a value
my_dict["country"] = "USA"  # Adding a new key-value pair
```

```python
myfamily = {
  "child1" : {
    "name" : "Emil",
    "year" : 2004
  },
  "child2" : {
    "name" : "Tobias",
    "year" : 2007
  },
  "child3" : {
    "name" : "Linus",
    "year" : 2011
  }
}

print(myfamily)
```

```python
print(myfamily["child2"]["name"])
```

```python
print(myfamily.keys())
```

# **loops**

```python
i = 1
while i < 6:
  print(i)
  i += 1
```

```python
#break Statement
i = 1
while i < 6:
  print(i)
  if i == 3:
    break
  i += 1
```

```python
#continue statement
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue
  print(i)
```

# **for loop**

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
for i in range(5):  # Loops from 0 to 4
    print(i)
```

```python
for char in "Python":
    print(char)
```
